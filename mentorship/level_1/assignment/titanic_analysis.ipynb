***1. Взять данные из файла.***

Для выполнения задания был выбран датасет "Титаник".

```python
file = pd.read_csv("titanic.csv")
```

***2.1. Первичная обработка данных***

Для первичной обработки данных узнаем размер взятого датасета, просмотрим первые строки общей таблицы, узнаем информацию о типах данных и количестве пропусков, найдём количество дубликатов и статистически опишем  имеющиеся числовые признаки из таблицы.

```python
print("Размер датасета: \n", file.shape)
print("\nПервые 5 строк: \n", file.head())
print("\nИнформация о типах данных и количестве пропусков: \n", file.info())
print("\nКоличество дубликатов: \n", file.duplicated().sum())
print("\nСтатистическое описание числовых признаков: \n", file.describe())
```

***2.2. Подсчёт простой статистики***

В данном блоке посчитаем общее количество пассажиров, количество выживших и погибших, а также вероятность выживания пассажира.
Для расчётов используем следующие формулы:
- **Количество погибших** = общее количество пассажиров − количество выживших
- **Вероятность выживания** = количество выживших / общее количество пассажиров
- **Доля выживших в процентах** = вероятность выживания * 100%

```python
# Кол-во выживших
survived_count = 0
for val in file["Survived"]:
    if val == 1:
        survived_count += 1

# Всего пассажиров
total_count = len(file["Survived"])

# Кол-во погибших
died_count = total_count - survived_count

# Вероятность выживания
survival_probability = survived_count / total_count
print(f"Всего пассажиров: {total_count}")
print(f"Выживших: {survived_count}")
print(f"Погибших: {died_count}")
print(f"Вероятность выживания: {survival_probability}")
print(f"Доля выживших: {survival_probability * 100} %")
```

В этом блоке подсчитаем математическое ожидание возраста выживших (среднее значение), дисперсию и среднеквадратическое отклонение. Для расчётов используем следующие формулы:
- **Мат. ожидание**: сумма всех возрастов / количество выживших
- **Дисперсия**: сумма значений, равных (x_i - x_среднее)^2
- **СКО**: корень из дисперсии

```python
# Математическое ожидание возраста выживших
survivors_age = file.loc[
    ((file["Survived"] == 1) & (file["Age"].notna())), "Age"
]
n = len(survivors_age)
age_expectation = sum(survivors_age) / n

# Дисперсия (разброс от среднего значения возраста выживших)
sum_squared_deviations = 0
for age in survivors_age:
    squared_deviation = (age - age_expectation) ** 2
    sum_squared_deviations += squared_deviation
age_dispersion = sum_squared_deviations / n

# Среднеквадратическое отклонение
stand_deviation = age_dispersion ** 0.5

print(f"Математическое ожидание возраста выживших: {age_expectation}")
print(f"Дисперсия возраста выживших: {age_dispersion}")
print(f"Среднеквадратическое отклонение возраста выживших: {stand_deviation}")
```

Далее проверим условную вероятность выживания по половому признаку. 

```python
women_total = 0
women_survived = 0

men_total = 0
men_survived = 0

for index, passenger in file.iterrows():
    if passenger["Sex"] == "female":
        women_total += 1
        if passenger["Survived"] == 1:
            women_survived += 1
    elif passenger["Sex"] == "male":
        men_total += 1
        if passenger["Survived"] == 1:
            men_survived += 1

# Подсчёт условной вероятности
p_survival_women = women_survived / women_total
p_survival_men = men_survived / men_total
print("Условная вероятность выживания по полу:")
print(f"Выжило женщин: {p_survival_women:.2%}")
print(f"Выжило мужчин: {p_survival_men:.2%}")
```
